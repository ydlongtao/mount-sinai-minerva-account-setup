# Visium HD LSF Monitor

This notebook is for monitoring LSF jobs, logs, and generated results. Do not run production analysis from this notebook.

```python
from pathlib import Path
import subprocess
import pandas as pd
from IPython.display import display, Markdown, Image, HTML

PROJECT_HOME = Path('/sc/arion/work/huangl21/sp_project')
RESULTS = PROJECT_HOME / 'results'
LOGS = PROJECT_HOME / 'logs'
print(PROJECT_HOME)
print(RESULTS.exists(), LOGS.exists())
```

## LSF Status

```python
def run(cmd):
    proc = subprocess.run(cmd, shell=True, text=True, capture_output=True)
    print(proc.stdout)
    if proc.stderr:
        print(proc.stderr)

run('bjobs -u $USER')
```

```python
# Replace JOBID with a real job id when needed.
JOBID = ''
if JOBID:
    run(f'bjobs -l {JOBID}')
else:
    print('Set JOBID to inspect a specific job.')
```

## Logs

```python
logs = sorted(LOGS.glob('*'))[-20:]
for p in logs:
    print(p.name, p.stat().st_size)
```

```python
# Set LOG_NAME to view a log file.
LOG_NAME = ''
if LOG_NAME:
    p = LOGS / LOG_NAME
    print(p)
    print(p.read_text(errors='replace')[-8000:])
else:
    print('Set LOG_NAME from the list above.')
```
